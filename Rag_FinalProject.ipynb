Text Preprocessing

Parse Markdown into Page Documents

```python
import re

with open(output_md, "r", encoding="utf-8") as f:
    markdown_text = f.read()

pages = re.split(r"## PDF Page \d+\n", markdown_text)

print("Number of extracted sections:", len(pages))
```

Clean and organize pages

```python
import re

# Split Markdown into individual PDF-page sections
pages = re.split(r"## PDF Page (\d+)\n", markdown_text)

page_documents = []

# pages structure:
# [text before first page, page_number, page_content, page_number, page_content, ...]
for i in range(1, len(pages), 2):
    page_number = int(pages[i])
    text = pages[i + 1].strip()

    # Clean whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Keep only pages that contain text
    if text:
        page_documents.append({
            "page_number": page_number,
            "content": text
        })

print("Total PDF pages:", 3623)
print("Pages containing text:", len(page_documents))
```

Book starting pages

```python
# First chapter title of each Harry Potter book
first_chapters = {
    "Harry Potter and the Sorcerer's Stone": "THE BOY WHO LIVED",
    "Harry Potter and the Chamber of Secrets": "THE WORST BIRTHDAY",
    "Harry Potter and the Prisoner of Azkaban": "OWL POST",
    "Harry Potter and the Goblet of Fire": "THE RIDDLE HOUSE",
    "Harry Potter and the Order of the Phoenix": "DUDLEY DEMENTED",
    "Harry Potter and the Half-Blood Prince": "THE OTHER MINISTER",
    "Harry Potter and the Deathly Hallows": "THE DARK LORD ASCENDING"
}

for book_name, chapter_title in first_chapters.items():
    matches = []

    for page_number, page in enumerate(doc, start=1):
        text = page.get_text("text").upper()

        if chapter_title in text:
            matches.append(page_number)

    print("=" * 70)
    print(book_name)
    print("First chapter:", chapter_title)
    print("Possible pages:", matches[:10])
```

Book Boundaries

```python
book_starts = {
    "Harry Potter and the Sorcerer's Stone": 12,
    "Harry Potter and the Chamber of Secrets": 282,
    "Harry Potter and the Prisoner of Azkaban": 573,
    "Harry Potter and the Goblet of Fire": 949,
    "Harry Potter and the Order of the Phoenix": 1570,
    "Harry Potter and the Half-Blood Prince": 2415,
    "Harry Potter and the Deathly Hallows": 2976
}

for book, start_page in book_starts.items():
    print(f"{book}: PDF page {start_page}")
```

Book Metadata

```python
# Assign each page to its corresponding book

book_ranges = list(book_starts.items())

for page in page_documents:
    page_number = page["page_number"]

    # Find the latest book whose starting page is <= current page
    current_book = None

    for i, (book_name, start_page) in enumerate(book_ranges):
        if page_number >= start_page:
            current_book = book_name

    page["book_name"] = current_book

print("Example page documents:")
for page in page_documents[:5]:
    print(page)
```

Verify book Metadata

```python
from collections import Counter

book_counts = Counter(
    page["book_name"]
    for page in page_documents
)

for book, count in book_counts.items():
    print(f"{book}: {count} pages")

print("\nPages without a book:", book_counts[None])
```

Chunking

```python
chunk_size = 500
chunk_overlap = 50

chunks = []

for page in page_documents:

    # Skip pages outside the actual seven books
    if page["book_name"] is None:
        continue

    # Skip the final copyright page
    if page["page_number"] == 3623:
        continue

    text = page["content"]
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk_text = text[start:end].strip()

        if chunk_text:
            chunks.append({
                "book_name": page["book_name"],
                "page_number": page["page_number"],
                "content": chunk_text
            })

        start += chunk_size - chunk_overlap

print("Total chunks:", len(chunks))

print("\nFirst chunk:")
print(chunks[0])

print("\nLast chunk:")
print(chunks[-1])
```

```python
# Inspect a few chunks from different parts of the dataset

for i in [0, 1, 100, 5000, 10000, len(chunks) - 1]:
    print("=" * 80)
    print(f"CHUNK {i}")
    print("=" * 80)
    print("Book:", chunks[i]["book_name"])
    print("PDF Page:", chunks[i]["page_number"])
    print("Content:", chunks[i]["content"][:500])
    print()
```

Load embedding model

```python
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("intfloat/multilingual-e5-large")

print("Embedding model loaded successfully!")
```

Generate embeddings

```python
texts = ["passage: " + chunk["content"] for chunk in chunks]

print("Number of texts to embed:", len(texts))
```

```python
batch_size = 32

embeddings = embedding_model.encode(
    texts,
    batch_size=batch_size,
    show_progress_bar=True,
    convert_to_numpy=True
)

print("Embeddings generated successfully!")
print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape[1])
```

Qdrant Vector Store

```python
!pip install -q qdrant-client
```

Create the Collection

```python
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams

qdrant_client = QdrantClient(":memory:")

collection_name = "harry_potter"

qdrant_client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(
        size=1024,
        distance=Distance.COSINE
    )
)

print("Qdrant collection created successfully!")
```

Upload vectors

```python
from qdrant_client.models import PointStruct

points = []

for i, chunk in enumerate(chunks):
    points.append(
        PointStruct(
            id=i,
            vector=embeddings[i].tolist(),
            payload={
                "book_name": chunk["book_name"],
                "page_number": chunk["page_number"],
                "content": chunk["content"]
            }
        )
    )

print("Prepared points:", len(points))
```

Upload vectors to Qdrant

```python
qdrant_client.upsert(
    collection_name=collection_name,
    points=points
)

print("Vectors uploaded successfully!")
```

Retrieval Testing

```python
query = "Who were Harry Potter's parents?"

query_embedding = embedding_model.encode(
    "query: " + query
)

print("Query embedding generated successfully!")
print("Vector size:", len(query_embedding))
```

Search Qdrant

```python
search_results = qdrant_client.query_points(
    collection_name=collection_name,
    query=query_embedding.tolist(),
    limit=3
).points

print("Retrieved results:", len(search_results))
```

Inspect Retrieved Results

```python
for i, result in enumerate(search_results, start=1):
    print("=" * 80)
    print(f"RESULT {i}")
    print("=" * 80)
    print("Similarity score:", result.score)
    print("Book:", result.payload["book_name"])
    print("PDF Page:", result.payload["page_number"])
    print("Content:", result.payload["content"])
    print()
```
